```python
import pandas as pd
df = pd.read_csv("../data/Fashion_Retail_Sales.csv")   # adapte au vrai nom
df.head()
```

```python
df_clean = df.copy()
df_clean.columns = ["client_id", "article", "montant", "date", "note", "paiement"]
```

```python
# Nettoyage
df_clean = df_clean[df_clean["montant"] > 0]
df_clean["date"] = pd.to_datetime(df_clean["date"], format="%d-%m-%Y")
df_clean["mois"] = df_clean["date"].dt.to_period("M").astype(str)
df_clean["jour_semaine"] = df_clean["date"].dt.day_name()
```

```python
# Enrichissement client (mini-RFM)
clients = df_clean.groupby("client_id").agg(
    nb_achats=("article", "count"),
    total_depense=("montant", "sum"),
    panier_moyen=("montant", "mean")
).reset_index()

df_final = df_clean.merge(clients, on="client_id", how="left")
df_final.to_csv("../data/fashion_tableau.csv", index=False)
print("Prêt pour Tableau :", df_final.shape)
```
